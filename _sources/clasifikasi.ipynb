# Pengertian Clasifikasi

Klasifikasi adalah cara membuat komputer memberi label pada sesuatu berdasarkan ciri-cirinya, dengan belajar dari contoh yang labelnya sudah diketahui.

Contoh sehari-hari: kamu menunjukkan 100 foto buah ke seorang anak dan memberi tahu mana apel dan mana jeruk. Anak itu belajar bahwa apel biasanya merah dan jeruk oranye. Setelah itu, saat melihat buah baru, ia bisa menebak jenisnya sendiri.

Klasifikasi di komputer sama persis:

Contoh yang sudah berlabel disebut data latih (training).
Ciri-ciri disebut fitur.
Cara komputer belajar disebut algoritma atau model (misalnya Random Forest).
Menebak data baru disebut prediksi.
Mengecek tebakan benar atau salah pada data yang tidak ikut belajar disebut data uji (testing).
Klasifikasi pada citra satelit

Citra satelit adalah kumpulan kotak kecil bernama piksel. Setiap piksel menyimpan angka seberapa banyak cahaya yang dipantulkan permukaan bumi pada tiap warna (band). Di sinilah ciri-cirinya:

| Permukaan | Ciri pantulannya |
|---|---|
| Tanaman hijau (sawah, lahan hijau, mangrove) | Menyerap cahaya merah, memantulkan banyak inframerah dekat |
| Air (laut, danau) | Hampir tidak memantulkan inframerah, tampak gelap |
| Bangunan | Memantulkan inframerah gelombang pendek (SWIR) lebih tinggi |

Jadi tiap piksel punya “sidik jari” angka. Klasifikasi tutupan lahan berarti: belajar sidik jari tiap jenis lahan dari contoh, lalu menebak jenis lahan untuk semua piksel lainnya. Hasilnya peta dengan warna berbeda untuk tiap kelas.
