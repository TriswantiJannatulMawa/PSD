# Pengantar Ekstraksi Data Kualitas Udara Kecamatan Kedungpring

## Kualitas Udara Kecamatan Kedungpring

Kecamatan Kedungpring dipilih sebagai wilayah kajian individu dalam proyek kelas ini,
sebagai bagian dari pembagian wilayah antar mahasiswa — setiap mahasiswa menganalisis
kecamatan yang berbeda yang kemudian dapat dibandingkan atau
digabungkan pada tahap analisis lanjutan (clustering).

## Profil Wilayah Kecamatan Kedungpring

Berdasarkan *Laporan Kinerja Instansi Pemerintah (LKjIP) Kecamatan Kedungpring 2024*
dan data administratif resmi:

| Atribut | Keterangan |
| ------- | ---------- |
| Luas wilayah | 84,54 km² |
| Jumlah desa | 23 desa |
| Batas Utara | Kecamatan Babat |
| Batas Timur | Kecamatan Sugio |
| Batas Selatan | Kecamatan Ngimbang |
| Batas Barat | Kecamatan Modo |
| Rentang koordinat | 112°10′01″–112°13′28″ BT, 07°08′19″–07°12′27″ LS |
| Titik pusat | 7°11′02″ LS, 112°12′15″ BT |

Kecamatan Kedungpring terletak di bagian barat Kabupaten Lamongan, berbatasan langsung
dengan beberapa kecamatan yang juga menjadi bagian dari kajian kelas ini. Wilayah ini
memiliki karakteristik campuran antara permukiman, pertanian, dan jalur penghubung
antar-kecamatan di sekitar koridor Babat–Jombang.

## Tujuan Ekstraksi Data

1. Menarik data konsentrasi polutan udara — **NO₂, CO, dan SO₂** — dari citra satelit
   **Sentinel-5P L2**, khusus untuk wilayah Kecamatan Kedungpring.
2. Membentuk deret waktu harian (*time series*) konsentrasi tiap polutan selama periode
   **31 Agustus 2025 – 31 Agustus 2026**.
3. Menyediakan data dasar yang bersih (bebas outlier dan missing value) untuk
   digunakan pada tahap **ekstraksi fitur (TSFEL)** dan **clustering** pada bagian
   proyek selanjutnya.

## Sumber & Metode Data

| Item | Keterangan |
| ---- | ---------- |
| Sumber data | Copernicus Data Space Ecosystem, layanan openEO |
| Produk satelit | Sentinel-5P L2 |
| Polutan | NO₂, CO, SO₂ |
| Periode waktu | 31 Agustus 2025 – 31 Agustus 2026 |
| Metode agregasi | Rata-rata harian (temporal) dan rata-rata dalam area kecamatan (spasial) |
| Area kajian (AOI) | Bounding box Kecamatan Kedungpring, dikonversi dari koordinat resmi LKjIP 2024 |
