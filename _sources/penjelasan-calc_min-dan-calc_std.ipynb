# Penjelasan dan Perhitungan dari fitur calc_min dan calc_std

## 1. Fitur calc_min

### 1.1 Pengertian 

calc_min adalah fitur statistik deskriptif dasar yang bertugas untuk mencari dan menampilkan nilai terkecil dari seluruh rangkaian data sensor polutan dalam satu periode pengamatan (window). Di dalam pustaka TSFEL, calc_min termasuk ke dalam kelompok Statistical Domain (Domain Statistik).

### 1.2 Manfaat dalam Analisis

- Mengetahui Titik Terendah Pencemaran: Membantu peneliti mengetahui kondisi paling bersih atau tingkat emisi terendah yang pernah tercatat oleh sensor di suatu wilayah (misalnya Kedungpring).

- Deteksi Anomali/Kesalahan Sensor: Jika nilai minimum bernilai negatif tidak wajar (seperti -0.000369 pada SO2), fitur ini langsung mendeteksi adanya offset atau kebutuhan kalibrasi ulang pada alat sensor.

- Baseline Konsentrasi: Berfungsi sebagai batas bawah untuk melihat rentang fluktuasi (range) kualitas udara bersamaan dengan nilai maksimumnya (calc_max).

### 1.3 Menghitung calc_min

#### 1.3.1 Calc_min SO2

Mencari nilai terkecil dari seluruh data yaitu 366 baris pada kolom SO2. Sehingga ditemukan nilai min nya **-0.0003685372357722** pada **7 april 2026**.

Keterangan: Pada hari tersebut, sensor merekam tingkat konsentrasi sulfur dioksida terendah yang mencatat angka negatif kecil (kemungkinan akibat penyesuaian kalibrasi atau baseline sensor).

#### 1.3.2 Calc_min CO

Mencari nilai terkecil dari seluruh data yaitu 366 baris pada kolom SO2. Sehingga ditemukan nilai min nya **0.020191341638565** pada **5 maret 2026**.

Keterangan: Pada tanggal ini, gas Karbon Monoksida di Kedungpring berada pada titik paling bersih atau emisi terendah sepanjang tahun pengamatan (berbeda dengan standar deviasinya yang tinggi, menunjukkan pada hari lain CO bisa melonjak tajam).

#### 1.3.3 Calc_min NO2

Mencari nilai terkecil dari seluruh data yaitu 366 baris pada kolom SO2. Sehingga ditemukan nilai min nya **7.337210263358429e-06** pada **8 april 2026**.

Keterangan: Menariknya, tanggal pencatatan nilai minimum NO2 ini berselang sehari setelah nilai minimum SO2 tercatat (7 April 2026), menunjukkan adanya tren kualitas udara yang sangat bersih pada awal bulan April 2026 di wilayah tersebut.

## 2. Fitur calc_std(standar deviasi)

### 2.1 Pengertian

calc_std adalah ukuran statistik yang menunjukkan seberapa jauh data tersebar dari nilai rata-ratanya ($\mu$). Jika grafiknya stabil datar-datar saja, maka nilai calc_std akan mendekati angka 0. Sebaliknya, jika dalam beberapa hari polusinya melonjak tinggi lalu anjlok drastis secara acak, nilai calc_std akan menjadi besar.

Sama seperti calc_min, calc_std juga masuk ke dalam kelompok Statistical Domain (Domain Statistik) di TSFEL, karena dihitung langsung menggunakan persamaan matematis standar deviasi terhadap seluruh titik data deret waktu.

### 2.2 Manfaat dalam Analisis

- Mengukur Tingkat Volatilitas/Fluktuasi: Membantu membandingkan tingkat kestabilan polutan. Sebagai contoh, dari data Anda:

    - CO memiliki calc_std paling tinggi (0.003424), yang berarti gas Karbon Monoksida di Kedungpring paling dinamis dan berfluktuasi (bisa naik/turun tajam, dipicu aktivitas harian seperti kendaraan).

    - SO2 dan NO2 memiliki calc_std yang sangat kecil (mendekati nol), menandakan konsentrasinya cenderung stabil dan datar sepanjang tahun.

- Indikator Stabilitas Lingkungan: Standar deviasi yang rendah sering diartikan bahwa kondisi lingkungan atau emisi sumber pencemar di daerah tersebut relatif konsisten dari waktu ke waktu.

### 1.4 Menghitung calc_std

#### 1.4.1 Calc_std SO2

- Menentukan nilai rata-rata dengan menjulahkan seluruh data lalu dibagi dengan N(366), Sehingga mendapatkan hasil $$\mu = 0.00005866$$

- Hitung selisih kuadrat dari rata-rata $\sum (x_i - \mu)^2$ , setiap data dikurangi dengan rata-rata lalu dikuadratkan dan setelah itu seluruh hasilnya dijumlahkan, sehingga hasilnya $8.68909 \times 10^{-6}$

- Hitung varians, $$\text{Varians} = \frac{\sum (x_i - \mu)^2}{N} = \frac{8.68909 \times 10^{-6}}{366} = \mathbf{2.37407 \times 10^{-8}}$$

- hitung standar deviasi, $$\sigma = \sqrt{\text{Varians}} = \sqrt{2.37407 \times 10^{-8}} = \mathbf{0.000154}$$

#### 1.4.2 calc_std CO

- Menghitung rata-rata, $$\mu = 0.02965457$$

- menghitung total kuadrat selisih, yaitu 0.00428976

- Menghitung varians, $$\text{Varians} = \frac{0.00428976}{366} = \mathbf{1.17206 \times 10^{-5}}$$

- Menghitung standar deviasi, $$\sigma = \sqrt{1.17206 \times 10^{-5}} = \mathbf{0.003424}$$

#### 1.4.3 Calc_std NO2

- Menghitung nilai rata-rata, $$\mu = 0.00002987$$

- Menghitung total kuadrat selisih,  $4.3550 \times 10^{-8}$

- Menghitung varians, $$\text{Varians} = \frac{4.3550 \times 10^{-8}}{366} = \mathbf{1.18989 \times 10^{-10}}$$

- Menghitung standar deviasi, $$\sigma = \sqrt{1.18989 \times 10^{-10}} = \mathbf{0.000011}$$
